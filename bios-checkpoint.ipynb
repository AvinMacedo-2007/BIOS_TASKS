```python
#Write a program to reverse only the alphabets in a string without changing the position of special characters.
n = int(input('Enter the number of elements in list: '))
l1 = []

for i in range(n):
    v = input(f'Enter element {i+1}: ')
    l1.append(v)

c1 = 0
c2 = len(l1) - 1

while c1 < c2:

    if not l1[c1].isalpha():
        c1 += 1
        continue

    if not l1[c2].isalpha():
        c2 -= 1
        continue

    # swap
    a = l1[c1]
    b = l1[c2]

    l1[c1] = b
    l1[c2] = a

    c1 += 1
    c2 -= 1

print(l1)
```

```python
#Write a function that takes two strings and returns True if they are anagrams of each other, and False otherwise.
n=input('enter first string')
m=input('enter second string')
def anagrams(n,m):
    if sorted(n)==sorted(m):
        return True
    else:
        return False
print(anagrams(n,m))
```
